# 1. Proses Scraping

```python
!pip install google-play-scraper

from google_play_scraper import reviews, Sort
import pandas as pd

# Contoh: aplikasi Gojek di Indonesia
result, _ = reviews(
    'org.telegram.messenger',          # app ID dari URL Play Store
    lang='id',
    country='id',
    sort=Sort.NEWEST,
    count=2000
)

data = [{
    'id': r['reviewId'],
    'teks': r['content'],
    'tanggal': r['at'],
    'sumber': 'playstore'
} for r in result]

df = pd.DataFrame(data)
df.to_csv('scraped_playstore_telegram.csv', index=False)
print('Total review:', len(df))
df.head()
```
