```python
import pandas as pd

df = pd.read_csv('/content/scraped_playstore_telegram.csv')   # ganti dengan file Anda
print('Dimensi data:', df.shape)
df.head()
```

```python
positif = [
    # Kata Umum, Singkat & Variasi Typo/Gaya Chat
    'bagus', 'mantap', 'top', 'oke', 'ok', 'baik', 'memuaskan', 'cocok', 'nyaman',
    'murah', 'suka', 'syukaaaa', 'love', 'best', 'hebat', 'keren', 'sukses', 'lancar',
    'aman', 'terbaik', 'josss', 'jos', 'cakep', 'nice', 'good', 'sipp', 'sip',
    'muantab', 'godd', 'goooddd', 'terbantu', 'membantu', 'bermanfaat', 'praktis',
    'ekonomis', 'terpercaya', 'jempol', 'bagusssss', 'gokil', 'rekomendasi', 'recsel',
    'works', 'worthit', 'mantul', 'mantap betul', 'recomended', 'rekomended', 'keren bgt',

    # Fitur Telegram & Kecepatan
    'ringan', 'kenceng', 'kencang', 'cepet', 'cepat', 'smooth', 'stabil', 'joss',
    'unlimited', 'gratis', 'bisa', 'berhasil', 'amanah', 'aman jaya', 'no kendala',
    'lancar jaya', 'anti lemot', 'update', 'fitur lengkap', 'canggih', 'mudah',
    'simpel', 'simple', 'fast respon', 'fast response', 'gercep'
]

negatif = [
    # Masalah Sistem, Server, & Fitur Telegram
    'buruk', 'jelek', 'lambat', 'lama', 'kecewa', 'rusak', 'mengecewakan', 'parah',
    'kapok', 'sampah', 'lemot', 'error', 'hilang', 'burik', 'bego', 'tolol', 'gajelas',
    'gak jelas', 'ga jelas', 'susah', 'lalot', 'pembodohan', 'haram', 'tidak bisa',
    'gak bisa', 'ga bisa', 'gabisa', 'kacau', 'payah', 'ngawur', 'membingungkan',
    'ribet', 'benci', 'beku', 'logout sendiri', 'suka eror', 'kurang bagus',
    'tidak sesuai', 'gak sesuai', 'ga sesuai', 'down', 'pending', 'connecting',
    'connecting terus', 'lag', 'nge-lag', 'ngelag', 'lemot banget', 'crash', 'force close',

    # Privasi, Keamanan, & Isu Komunitas (Spam/Scam/Banned)
    'scam', 'penipuan', 'penipu', 'ditipu', 'tipu', 'spammer', 'spam', 'banned',
    'terblokir', 'blokir', 'di-banned', 'dibanned', 'kena ban', 'hack', 'dihack',
    'di-hack', 'bajak', 'dibajak', 'akun hilang', 'dihapus', 'otp lama', 'gak masuk',
    'tidak masuk', 'gagal', 'gagal kirim', 'palsu', 'fake', 'bot jelek', 'bot eror',
    'virus', 'bahaya', 'rugi', 'saldo hilang', 'bikin emosi', 'jengkel', 'kecewa berat'
]

print('Jumlah kata positif:', len(positif))
print('Jumlah kata negatif:', len(negatif))
```

```python
def auto_label(teks):
    if not isinstance(teks, str):
        return 'netral'
    t = teks.lower()
    skor_pos = sum(1 for k in positif if k in t)
    skor_neg = sum(1 for k in negatif if k in t)

    if skor_pos > skor_neg:
        return 'positif'
    elif skor_neg > skor_pos:
        return 'negatif'
    else:
        return 'netral'

df['sentimen_auto'] = df['teks'].apply(auto_label)
print(df['sentimen_auto'].value_counts())
df.head(10)
```

```python
# Setelah dikoreksi di Excel/Sheets, simpan kembali sebagai 'sudah_dikoreksi.csv' lalu muat:
koreksi = pd.read_csv('sudah_dikoreksi.csv')
```

```python
koreksi.info()
```

```python
print(df.columns.tolist())
print(koreksi.columns.tolist())
```

```python
# ============================================================
# 1. Buat mapping hasil koreksi berdasarkan ID
# ============================================================

mapping_koreksi = (
    koreksi
    .dropna(subset=['koreksi'])
    .set_index('id')['koreksi']
)


# ============================================================
# 2. Buat label final
#    Jika ada koreksi -> gunakan koreksi
#    Jika tidak ada koreksi -> gunakan sentimen_auto
# ============================================================

df['sentimen'] = (
    df['id']
    .map(mapping_koreksi)
    .fillna(df['sentimen_auto'])
)


# ============================================================
# 3. Cek hasil 50 data yang dikoreksi
# ============================================================

print("50 data setelah koreksi:")
print(
    df.head(50)[
        ['id', 'sentimen_auto', 'sentimen']
    ]
)


# ============================================================
# 4. Buang data selain positif dan negatif
# ============================================================

df_2kelas = df[
    df['sentimen'].isin(['positif', 'negatif'])
].reset_index(drop=True)


# ============================================================
# 5. Simpan dataset final
# ============================================================

df_2kelas.to_csv('data_labeled.csv', index=False)


# ============================================================
# 6. Cek distribusi sentimen final
# ============================================================

print("\nDistribusi sentimen final:")
print(df_2kelas['sentimen'].value_counts())
```
